```python
import numpy as np
from pathlib import Path

features_path = Path(
    "/content/drive/MyDrive/Paddy_Disease_Project/features"
)

X_train = np.load(features_path / "X_train.npy")
y_train = np.load(features_path / "y_train.npy")
X_valid = np.load(features_path / "X_valid.npy")
y_valid = np.load(features_path / "y_valid.npy")

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_valid:", X_valid.shape)
print("y_valid:", y_valid.shape)
```

```python
classes = [
    'bacterial_leaf_blight',
    'bacterial_leaf_streak',
    'bacterial_panicle_blight',
    'blast',
    'brown_spot',
    'dead_heart',
    'downy_mildew',
    'hispa',
    'normal',
    'tungro'
]
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report
import time

knn_model = KNeighborsClassifier(
    n_neighbors=5,
    weights='distance',
    metric='euclidean',
    n_jobs=-1
)

# Training
start_time = time.time()
knn_model.fit(X_train, y_train)
knn_train_time = time.time() - start_time

# Prediction
start_time = time.time()
knn_preds = knn_model.predict(X_valid)
knn_inference_time = time.time() - start_time

# Evaluation
knn_accuracy = accuracy_score(y_valid, knn_preds)

print(f"KNN Training time: {knn_train_time:.4f} seconds")
print(f"KNN Inference time: {knn_inference_time:.4f} seconds")
print(f"\nKNN Accuracy: {knn_accuracy * 100:.2f}%")

print("\nClassification Report:")
print(classification_report(
    y_valid,
    knn_preds,
    target_names=classes,
    digits=4
))
```

```python
import joblib

model_path = "/content/drive/MyDrive/Paddy_Disease_Project/models"

joblib.dump(
    knn_model,
    f"{model_path}/model_3_knn.pkl"
)

print("KNN model saved successfully.")
```
